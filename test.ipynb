```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-10, 10)
x = np.where(n == 0, 1, 0)

plt.stem(n, x)
plt.title("Unit Sample Sequence δ(n)")
plt.xlabel("n")
plt.ylabel("δ(n)")
plt.grid(True)
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-10, 10)
x = np.where(n >= 0, 1, 0)

plt.stem(n, x)
plt.title("unit step sequence u(n)")
plt.xlabel("time")
plt.ylabel("amplitude")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-10, 11)
a_values = [0.5, 1.0, 1.5, -0.5]

plt.figure(figsize=(10, 8))
for i, a in enumerate(a_values, 1):
    x = np.where(n >= 0, a ** n, 0)
    plt.subplot(2, 2, i)
    plt.stem(n, x, basefmt=" ")
    plt.title(f"Exponential a^n u(n), a = {a}")
    plt.xlabel("n")
    plt.ylabel("x(n)")
    plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pylab as plt

n = np.arange(-10, 11)

def x_base(n):
    return np.where((n >= 0) & (n <= 4), n + 1, 0)


plt.figure(figsize=(10, 8))

plt.subplot(3, 1, 1)
plt.stem(n, x_base(n))
plt.title("Unit Ramp Sequence")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)


plt.subplot(3, 2, 1)
plt.stem(n, x_base(n) - 2)
plt.title("advance (n - 2)")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)


plt.tight_layout()
plt.show()

```
